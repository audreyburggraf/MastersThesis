```python
# Load the neural network
nndir = '../nnmodels' # Change this path
producer = glitterin.user.ScatteringProducer(nndir=nndir)

# Now we want to decide what qantities we want. 
producer.setup(['Cext', 'albedo', 'Z11'])
```

```python
# The inputs (x,n,k) have to be matching 1d arrays. 
# theta should be its own 1d array. 
xenc = np.array([10])
re_m = np.array([1.5])
im_m = np.array([1e-2])

theta = np.linspace(0, 180, 181)

# The wavelength can be a float or an array that matches the length of (x,n,k)
w = 1e-4 # [cm]
```

```python
# Call the producer
out = producer(xenc, re_m, im_m, theta, w, xtype='enc', outtype='MuellerMatrix')

# Make things easier. Let's just consider the angular dependence
Z11 = out.Z11[:,0]
```

```python
# Calculate the asymmetry parameter through the gridded approach

# Define the mu values at the walls
mu = np.cos(theta * np.pi / 180)

# Define the mu values at the cells
mu_c = 0.5 * (mu[1:] + mu[:-1])

# Define the step size
d_mu = (mu[1:] - mu[:-1])

# Also define Z11 at the cells
Z11_c = 0.5 * (Z11[1:] + Z11[:-1])

# Numerator
int_Z11_mu_dmu = np.sum(Z11_c * mu_c * d_mu)

# Denominator
int_Z11_dmu = np.sum(Z11_c * d_mu)

g = int_Z11_mu_dmu / int_Z11_dmu
```
