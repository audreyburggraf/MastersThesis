# Set my bands and run

```python
# Constants
bands = ["Band 4 nterms2", "Band 5", "Band 6", "Band 5 robust -1", "Band 7 nterms2"]
bands_naming = ['Band4_nterms2', 'Band5', 'Band6', "Band5_robust_minus1", 'Band7_nterms2']
# bands_labels = ['Band 4', 'Band 5', 'Band 6', 'Band 7']

num_bands = len(bands)
 
lambda_bands_cm = mm_to_cm([lambda_mm[b] for b in bands])
lambda_bands_micron = mm_to_micron([lambda_mm[b] for b in bands])
band_colors = [alma_band_colors[b] for b in bands]

oc, rho_g_cm3 = do.get_dsharp_mix()
```

```python
data_n = {}
data_k = {}

for b, lam_cm in zip(bands, lambda_bands_cm):
    nk = oc.nk(lam_cm)
    
    data_n[b] = nk[0]
    data_k[b] = nk[1]

    print(b)
    print("lambda =", lam_cm, "cm")
    print("n =", data_n[b])
    print("k =", data_k[b])
    print()
```

```python
import numpy as np

# Build rows: band, lambda_cm, n, k
rows = []

for b, idx, lam_cm in zip(bands, bands_naming, lambda_bands_cm):
    rows.append([idx, lam_cm, data_n[b], data_k[b]])

rows = np.array(rows, dtype=object)

np.savetxt(
    "nk_values.txt",
    rows,
    fmt="%s",
    header="band lambda_cm n k",
    comments=""
)
```
