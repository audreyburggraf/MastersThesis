```python
band = B4
theta_rad = constants.major_angle_rad_cartesian_band4_nterms2
band_idx = 'Band 4 nterms2'
step = constants.constants.step_band4_nterms2



fig_x_size = 14
fig_y_size = 12

cmap=soft_colormap_no_red

vmin = 0 
vmax = 1




custom_phi  = [3, 7]
custom_BMAJ = [40, 50]
custom_BMIN = [30, 20]


fig = plt.figure(figsize=(fig_x_size, fig_y_size))
gs = GridSpec(1, 2, figure=fig, wspace = 0.05)

axes = []

saving_ratios = []

for i in range(2):
    # Set the phi, BMAJ, BMIN values
    
    phi = custom_phi[i]
    BMAJ = custom_BMAJ[i]
    BMIN = custom_BMIN[i]
    
    print(rf'i = {i} and phi = {phi} and BMAJ = {BMAJ} and BMIN = {BMIN}')

    GaussianUniformRatios, GaussianAzimuthalRatios, _, _ = gaussian_2d_flat_topped_tilted_model(band['nx'], 
                                                                                                band['ny'], 
                                                                                                theta_rad, 
                                                                                                phi, 
                                                                                                BMIN,
                                                                                                BMAJ, 
                                                                                                band['RA_centre_pix'],
                                                                                                band['Dec_centre_pix'])
    
    
    _, _, _, vectors_data, vectors_angle, _  = mix_StokesQU_and_generate_vectors(GaussianUniformRatios,
                                                                         GaussianAzimuthalRatios, 
                                                                         StokesQ_grid_100Uniform,
                                                                         StokesU_grid_100Uniform,
                                                                         StokesQ_grid_100Azimuthal, 
                                                                         StokesU_grid_100Azimuthal,
                                                                         band['ny'], band['nx'], 
                                                                         band['StokesI_mJy'], band['StokesI_err_mJy'], 
                                                                         band['POLI_debiased_mJy'], band['POLI_err_mJy'], 
                                                                         band['PA_err_deg'],
                                                                         band_idx,
                                                                         step)

        
    
    
    
    saving_ratios.append(GaussianUniformRatios)
    
    
    
    
    row = i // 2
    col = i % 2

    ax = fig.add_subplot(gs[row, col], projection=band['StokesI_wcs'])
    axes.append(ax)

    im = ax.imshow(
        GaussianUniformRatios,
        origin='lower',
        cmap=cmap,
        vmin=vmin,
        vmax=vmax
    )
    
    for row in vectors_data:
        ax.plot([row[0], row[1]], [row[2], row[3]], color='black', lw=writeup_grid_lw)

    ax.set_xlim(band['xmin'], band['xmax'])
    ax.set_ylim(band['ymin'], band['ymax'])
    
    print(rf'({xmin}, {xmax}), ({ymin}, {ymax})')


    ax.text(
    0.05, 0.25,
    rf"$\phi={phi}$" + "\n" +
    rf"$a={BMAJ}$" + "\n" +
    rf"$b$ $={BMIN}$" + "\n" ,
#     rf"$\chi^2={band_data['chi2']:.3f}$",
    transform=ax.transAxes,
    fontsize=constants.text_fs * 0.7,
    color='black',
    va='top',
    ha='left',
#         bbox=dict(facecolor='white', alpha=0.7, edgecolor='none')
)



    ax.set_xlabel(" ", fontsize= 1)
    ax.set_ylabel(" ", fontsize= 1)

    ax.tick_params(axis="x", which="major",
                   direction="in",
                   bottom=True, top=True,
                   length=7,
                   labelbottom=False)

    ax.tick_params(axis="y", which="major",
                   direction="in",
                   left=True, right=True,
                   length=7,
                   labelleft=False)

# One shared colour bar
cbar = fig.colorbar(im, ax=axes, orientation='vertical', 
                    fraction=0.045, pad=0.04, shrink = 0.55)


cbar.set_label(r"$\mathrm{Gaussian\ Uniform\ Weights}\ (W_{\mathrm{Uniform}})$", fontsize=constants.cbar_fs,
              labelpad = 15)
cbar.ax.tick_params(labelsize=constants.cbar_num_fs)

# Add the wavelength label
add_band_label(axes[0], 4,  'wavelength', constants)

# Test differences
print(' ')
print(rf'Top left and Top right: {np.nansum(saving_ratios[0] - saving_ratios[1])}')

plt.savefig(writeup_image_folder_path + '/Gaussian/' + "CustomExamples_B4.pdf", 
            dpi=300, 
            bbox_inches='tight')  
```
