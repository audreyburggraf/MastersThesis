```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 3, 500)   # x = a / lambda

# Scattering efficiency (sigmoid)
sca = 1 / (1 + np.exp(-5*(x-1)))

# Polarization (Gaussian)
pol = np.exp(-((x-0.6)/0.35)**2)

plt.figure(figsize=(6,4))

plt.plot(x, sca, lw=3, label='Scattering efficiency')
plt.plot(x, pol, lw=3, label='Polarization')

plt.xlabel(r'Grain size, $a/\lambda$')
plt.ylabel('Relative value')

plt.xlim(0,3)
plt.ylim(0,1.1)

plt.xticks([0,0.5,1,2,3])
plt.yticks([])

plt.legend(frameon=False)

plt.tight_layout()
plt.show()
```
