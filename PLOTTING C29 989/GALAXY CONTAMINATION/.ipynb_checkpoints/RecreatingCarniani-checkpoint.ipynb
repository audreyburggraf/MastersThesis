### Schechter

Equation (4):


$N(>S) dS = N_0 \left(\frac{S}{S_0}\right)^\alpha \exp \left(\frac{S}{S_0}\right) d \left(\frac{S}{S_0}\right)$

```python
def Schechter(params, S):
    alpha, S0, N0 = params
    return (N0/S0) * (S/S0)**alpha * np.exp(-S/S0)
```

### Double power-law

Equation (5):

$N(>S) dS = N_0 \left[\left(\frac{S}{S_0}\right)^\alpha + \left(\frac{S}{S_0}\right)^\beta \right]^{-1} d \left(\frac{S}{S_0}\right)$

## Do the fit

least_squares reads in:
 - function
 - x0: arrap of initial guess on independent variables
- args: the x and y data 


The guesses will be the results from the paper

```python
Guess = [-2.08, 1.7, 1800]
```

```python
results = least_squares(log_residual, Guess,
                        args=(Schechter, df["S"].values, df["dN/dS"].values))
```

```python

S_grid = np.logspace(np.log10(df["S"].min()/2),
                     np.log10(df["S"].max()*2),
                     500)
```

```python
results.x
```

```python
schechter_fit = Schechter([-7.81908642e-01, 1.63461236e+00,  1.77221100e+03], S_grid)


schechter_fit_their_values = Schechter([-2.08, 1.7, 1800], S_grid)
```
