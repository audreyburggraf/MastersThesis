Equation (5):

$N(>S) dS = N_0 \left[\left(\frac{S}{S_0}\right)^\alpha + \left(\frac{S}{S_0}\right)^\beta \right]^{-1} d \left(\frac{S}{S_0}\right)$

```python
# def dpl_diff(S, N0, S0, alpha, beta):
#     """
#     Differential Double Power Law (DPL).
    
#     Parameters:
#         S   : flux density [mJy]
#         N0  : normalization [mJy^-1 deg^-2]
#         S0  : break flux density [mJy]
#         alpha, beta : power-law slopes
#     """
#     return N0 * ((S / S0)**alpha + (S / S0)**beta)**-1
```

```python
from scipy.special import gammaincc, gamma
from scipy.integrate import quad

def schechter_diff(params, S):
    alpha, S0, N0 = params
    return N0 * (S/S0)**alpha * np.exp(-S/S0)

def dpl_diff(params, S):
    alpha, beta, S0, N0 = params
    return N0 / ((S/S0)**alpha + (S/S0)**beta)

def schechter_cum(params, S):
    alpha, S0, N0 = params
    Sgrid = np.linspace(S, 100, 1000)   # integrate from S to 100 mJy
    return np.trapz(schechter_diff([alpha,S0,N0], Sgrid), Sgrid)

def dpl_cum(params, S):
    alpha, beta, S0, N0 = params
    Sgrid = np.linspace(S, 100, 1000)
    return np.trapz(dpl_diff([alpha,beta,S0,N0], Sgrid), Sgrid)


def log_residual(params, func, x, y):
    # pass params as a single array (do NOT unpack)
    model = np.array([func(params, xi) for xi in x])
    mask = (model > 0) & (y > 0)
    return np.log10(model[mask]) - np.log10(y[mask])
```

# Run least square fits

least_squares reads in:
 - function
 - x0: arrap of initial guess on independent variables
- args: the x and y data 


The guesses will be the results from the paper

```python
# Differential Schechter (alpha, S0, N0)
diff_sch_Chen = [-1.7, 3.9, 2.2]

# Differential DPL (alpha, beta, S0, N0)
diff_dpl_Chen = [2.0, 4.7, 5.6, 0.8]

# Cumulative Schechter (alpha, S0, N0)
cum_sch_Chen = [-0.6, 2.2, 9.6]

# Cumulative DPL (alpha, beta, S0, N0)
cum_dpl_Chen = [-1.5, -3.0, 1.0, 1e3]
```

```python

S_mJy = np.array([0.25, 0.42, 0.69, 1.15, 1.90, 3.14, 5.19])   # midpoints for differential
dN_dS = np.array([8.0, 5.0, 4.0, 3.0, 2.0, 1.5, 1.0])          # example counts

S_bins_cumulative = np.array([0.19, 0.32, 0.52, 0.86, 1.43, 2.36, 3.91, 6.47])
N_cum = np.array([20, 15, 12, 8, 6, 4, 2, 1])  


# 1. Differential Schechter
res_sch_diff = least_squares(log_residual, diff_sch_Chen,
                             args=(schechter_diff, S_mJy, dN_dS))
best_fit_params_sch_diff = res_sch_diff.x

# 2. Differential DPL
res_dpl_diff = least_squares(log_residual, diff_dpl_Chen,
                             args=(dpl_diff, S_mJy, dN_dS))
best_fit_params_dpl_diff = res_dpl_diff.x

# 3. Cumulative Schechter
res_sch_cum = least_squares(log_residual, cum_sch_Chen,
                            args=(schechter_cum, S_bins_cumulative, N_cum))
best_fit_params_sch_cum = res_sch_cum.x

# 4. Cumulative DPL
res_dpl_cum = least_squares(log_residual, cum_dpl_Chen,
                            args=(dpl_cum, S_bins_cumulative, N_cum))
best_fit_params_dpl_cum = res_dpl_cum.x


# Print resilts



print("\n===== FIT RESULTS =====")
print("Differential Schechter: alpha = %.3f, S0 = %.3g, N0 = %.3g" %
      tuple(best_fit_params_sch_diff))
print("Differential DPL:       alpha = %.3f, beta = %.3f, S0 = %.3g, N0 = %.3g" %
      tuple(best_fit_params_dpl_diff))
print("Cumulative Schechter:   alpha = %.3f, S0 = %.3g, N0 = %.3g" %
      tuple(best_fit_params_sch_cum))
print("Cumulative DPL:         alpha = %.3f, beta = %.3f, S0 = %.3g, N0 = %.3g" %
      tuple(best_fit_params_dpl_cum))
```

# Plots (Figure 13)

```python
# Make smooth flux grids for plotting the fits
S_plot_diff = np.logspace(np.log10(0.1), np.log10(10), 500)      # differential
S_plot_cum  = np.logspace(np.log10(0.05), np.log10(50), 500)     # cumulative

# Evaluate the fits
y_sch_diff = np.array([schechter_diff(best_fit_params_sch_diff, s) for s in S_plot_diff])
y_dpl_diff = np.array([dpl_diff(best_fit_params_dpl_diff, s) for s in S_plot_diff])

y_sch_cum = np.array([schechter_cum(best_fit_params_sch_cum, s) for s in S_plot_cum])
y_dpl_cum = np.array([dpl_cum(best_fit_params_dpl_cum, s) for s in S_plot_cum])
```
