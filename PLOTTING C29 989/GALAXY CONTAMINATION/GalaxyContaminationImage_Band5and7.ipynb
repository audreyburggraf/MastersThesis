# Plot both images together

```python
def MakeGalaxyContaminationPlot(B5, B7,
                                fig_x_size = 28, 
                                fig_y_size = 12,
                                axis_num_fs = constants.axis_num_fs,
                                cbar_fs = constants.cbar_fs,
                                cmap = "afmhot",
                                fontcolor = 'white',
                                reference_lw = 3,
                                beam_lw = 2,
                                spline_lw = 3,
                                label = 'wavelength'):
    
    # StokeI_wcs will be the same for boht B5 and B7
    StokesI_wcs = B5_plotting['StokesI_wcs']
    
    
    # Make the plots
    fig, ax = plt.subplots(1, 2, figsize=(fig_x_size, fig_y_size),
                       gridspec_kw={'wspace': -0.5},  # Increase wspace for more horizontal space
                       subplot_kw={'projection': StokesI_wcs},
                      sharey = True)
    
    
    # Set what we are plotting
    # ------------------------------------------------------------------
    im0 = ax[0].imshow(B5['StokesI_stretched'], cmap = cmap)
    im1 = ax[1].imshow(B7['StokesI_stretched'], cmap = cmap)
    # ------------------------------------------------------------------
    
    
    # Colourbars
    # ------------------------------------------------------------------
    # Colorbar for Band 5 
    cbar0 = add_StokesI_cbar(
        im0,
        ax[0],
        B5['normalized_cbar_ticks'],
        B5['StokesI_unstretched_cbar_ticks'],
        cbar_fs,
        axis_num_fs,
        shrink = 0.4
    )
    # ------------------------------------------------------------------
    # Colorbar for Band 7
    cbar1 = add_StokesI_cbar(
        im1,
        ax[1],
        B7['normalized_cbar_ticks'],
        B7['StokesI_unstretched_cbar_ticks'],
        cbar_fs,
        axis_num_fs,
        shrink = 0.4
    )
    # ------------------------------------------------------------------
    
    
    # Add axis labels
    # ------------------------------------------------------------------
    ax[0].set_xlabel('Right Ascension', fontsize=axis_label_fs)
    ax[1].set_xlabel('Right Ascension', fontsize=axis_label_fs)
    ax[0].set_ylabel('Declination', fontsize=axis_label_fs)
    ax[1].coords[1].set_ticklabel_visible(False)
    ax[1].coords[1].set_axislabel('')
    # ------------------------------------------------------------------
    
    # Add line and text for 100 AU 
    # ------------------------------------------------------------------
    add_reference_scale_bar(ax[0], 
                            B5['xmin'], B5['xmax'], B5['ymin'], B5['ymax'],
                            B5['reference_length_pix'],
                            B5['reference_length_AU'],
                            text_fs, fontcolor, reference_lw)
    
    add_reference_scale_bar(ax[1], 
                            B7['xmin'], B7['xmax'], B7['ymin'], B7['ymax'],
                            B7['reference_length_pix'],
                            B7['reference_length_AU'],
                            text_fs, fontcolor, reference_lw)
    # ------------------------------------------------------------------
    
    # Add the beams
    # ------------------------------------------------------------------
    # Create and add the circular beam
    beam_x_pos = B5['xmin'] - 0.1 * (B5['xmin'] - B5['xmax'])
    beam_y_pos = B5['ymin'] - 0.1 * (B5['ymin'] - B5['ymax'])

    beam = Ellipse((beam_x_pos, beam_y_pos), 
                   width = B5['BMAJ_pix'], height = B5['BMIN_pix'],  
                   angle = B5['BPA_deg_cartesian'], 
                   edgecolor= fontcolor, 
                   facecolor='none', 
                   alpha=1, 
                   lw=beam_lw)

    ax[0].add_patch(beam)
    # ------------------------------------------------------------------
    beam_x_pos = B7['xmin'] - 0.1 * (B7['xmin'] - B7['xmax'])
    beam_y_pos = B7['ymin'] - 0.1 * (B7['ymin'] - B7['ymax'])

    beam = Ellipse((beam_x_pos, beam_y_pos), 
                   width = B7['BMAJ_pix'], height = B7['BMIN_pix'],  
                   angle = B7['BPA_deg_cartesian'], 
                   edgecolor= fontcolor, 
                   facecolor='none', 
                   alpha=1, 
                   lw=beam_lw)

    ax[1].add_patch(beam)
    # ------------------------------------------------------------------
    
    
    # Formatting axes
    # ------------------------------------------------------------------
    add_min_major_ticks_wcs(ax[0])
    add_min_major_ticks_wcs(ax[1])
    # ------------------------------------------------------------------
    # Make the axes thicker
    ax[0].coords.frame.set_linewidth(spline_lw)
    ax[1].coords.frame.set_linewidth(spline_lw)
    
    ax[0].tick_params(axis="both", which="major", width=spline_lw)
    ax[1].tick_params(axis="both", which="major", width=spline_lw)
    # ------------------------------------------------------------------
        
        
    # Add the band labels
    # ------------------------------------------------------------------
    add_band_label(ax[0], 5, label, text_fs, fontcolor, constants)
    add_band_label(ax[1], 7, label, text_fs, fontcolor, constants)
    # ------------------------------------------------------------------
    
    
        
    return fig, ax
```

```python
B5_plotting = {
    'StokesI_wcs': StokesI_wcs_B5,
    'StokesI_stretched': StokesI_stretched_mJy_B5,
    'normalized_cbar_ticks': normalized_cbar_ticks_B5,
    'StokesI_unstretched_cbar_ticks': StokesI_unstretched_cbar_ticks_B5, 
    'xmin': 0,
    'xmax': nx_B5,
    'ymin': 0,
    'ymax': ny_B5,
    'reference_length_pix': reference_length_pix_B5,
    'reference_length_AU': reference_length_AU_B5,
    'BMIN_pix': BMIN_pix_B5,
    'BMAJ_pix': BMAJ_pix_B5,
    'BPA_deg_cartesian':BPA_deg_cartesian_B5,
}

B7_plotting = {
    'StokesI_wcs': StokesI_wcs_B7,
    'StokesI_stretched': StokesI_stretched_mJy_B7,
    'normalized_cbar_ticks': normalized_cbar_ticks_B7,
    'StokesI_unstretched_cbar_ticks': StokesI_unstretched_cbar_ticks_B7, 
    'xmin': 0,
    'xmax': nx_B7,
    'ymin': 0,
    'ymax': ny_B7,
    'reference_length_pix': reference_length_pix_B7,
    'reference_length_AU': reference_length_AU_B7,
    'BMIN_pix': BMIN_pix_B7,
    'BMAJ_pix': BMAJ_pix_B7,
    'BPA_deg_cartesian':BPA_deg_cartesian_B7,
}



fig, ax = MakeGalaxyContaminationPlot(B5_plotting, B7_plotting)


plt.savefig(writeup_image_folder_path + '/GalaxyContamination/'+ "StokesI_BothBands.pdf", 
            dpi=300, 
            bbox_inches='tight')   
```
