```python
def get_arrays(nterms1_data, nterms2_data):

    PA_nterms1, POLF_nterms1, StokesI_nterms1, vector_mask_nterms1, _ = nterms1_data
    PA_nterms2, POLF_nterms2, StokesI_nterms2, vector_mask_nterms2, _ = nterms2_data
    
    # Make another mask where both masks are true 
    common_mask = vector_mask_nterms1 & vector_mask_nterms2
    

    
    # Make the arrays for the data in the plot 
    POLF_nterms1 = POLF_nterms1[common_mask]
    POLF_nterms2 = POLF_nterms2[common_mask]
    
    PA_nterms1   = PA_nterms1[common_mask]
    PA_nterms2   = PA_nterms2[common_mask]
    
    StokesI_nterms1   = StokesI_nterms1[common_mask]
    StokesI_nterms2   = StokesI_nterms2[common_mask]
    
#     StokesQ_nterms1   = StokesQ_nterms1[common_mask]
#     StokesQ_nterms2   = StokesQ_nterms2[common_mask]
    
#     StokesU_nterms1   = StokesU_nterms1[common_mask]
#     StokesU_nterms2   = StokesU_nterms2[common_mask]
                                   
    
    data = [PA_nterms1, PA_nterms2,
            POLF_nterms1, POLF_nterms2, 
            StokesI_nterms1, StokesI_nterms2]
#             StokesQ_nterms1, StokesQ_nterms2,
#             StokesU_nterms1, StokesU_nterms2]
    
    return data
```

```python
def plot_unity(ax, xdata, ydata):
    lo = min(xdata.min(), ydata.min())
    hi = max(xdata.max(), ydata.max())
    u = np.linspace(lo, hi, 100)
    ax.plot(u, u, color='black', lw=5, ls='--', label='Unity')
    

```
