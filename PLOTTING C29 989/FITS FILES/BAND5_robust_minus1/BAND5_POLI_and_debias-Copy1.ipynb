```python
StokesI_file = path + "IRS63_BAND5_StokesI_redo_nterms1.fits"
StokesQ_file = path + "IRS63_BAND5_StokesQ_redo_nterms1.fits"
StokesU_file = path + "IRS63_BAND5_StokesU_redo_nterms1.fits"
```

```python
# Stokes I
# -------------------------------------------------------------------------------------------------------
StokesI_header, _, StokesI_Jy, StokesI_wcs = read_in_file(StokesI_file)
StokesI_mJy = convert_jy_to_mjy(StokesI_Jy)
nx, ny = StokesI_mJy.shape
```
