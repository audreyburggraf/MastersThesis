```python
csv_files_v4 = {
    1:    "data_debugging_v4_1_ZHANG.csv",
    0.3: "data_debugging_v4_0_3_ZHANG.csv",
    0.1:  "data_debugging_v4_0_1_ZHANG.csv",
    0.01: "data_debugging_v4_0_01_ZHANG.csv",
}
```

```python
# Make some empty arrays for stuff

a_multiplied_by_f_dist_micron_v4 = {}
P_times_omega_v4 = {}
P_times_omega_eff_v4 = {}
P_v4 = {}
w_v4 = {}
w_eff_v4 = {}



POLF_obs_v4 = {}



# Loop over f values and csv files
for f, fname in csv_files_v4.items():
    

    df = pd.read_csv(path_v4  + fname)

    # x-axis
    a_max_multiplied_by_f_vals = df["Xdata"].values
    a_multiplied_by_f_dist_micron_v4[f] = a_max_multiplied_by_f_vals

    # ALL bands
    Pw = np.column_stack([
        df[f"P_times_omega_{b}"].values for b in bands_naming
    ])
    
    Pw_eff = np.column_stack([
        df[f"P_times_omega_eff_{b}"].values for b in bands_naming
    ])
    
    P = np.column_stack([
        df[f"P_{b}"].values for b in bands_naming
    ])
        
    w = np.column_stack([
        df[f"omega_{b}"].values for b in bands_naming
    ])
    
    
    w_eff = np.column_stack([
        df[f"omega_eff_{b}"].values for b in bands_naming
    ])
    
    
            
            
    P_times_omega_v4[f] = Pw
    P_v4[f] = P
    w_v4[f] = w
    w_eff_v4[f] = w_eff
    P_times_omega_eff_v4[f] = Pw_eff
    
```

```python
for f in f_values:
    minn = micron_to_cm(min(a_multiplied_by_f_dist_micron_v4[f]))
    maxx = micron_to_cm(max(a_multiplied_by_f_dist_micron_v4[f]))
    print(rf'f = {f}, min = {minn}, max = {maxx}')
```
