```python
def save_results(location, name, results, single_wavelength=True):

    if single_wavelength:
        k_abs = results['k_abs'][:, 0]
        k_sca = results['k_sca'][:, 0]
        g = results['g'][:, 0]
        S1 = results['S1'][:, 0, :]
        S2 = results['S2'][:, 0, :]
        lam = results['lam'][0]
    else:
        k_abs = results['k_abs']
        k_sca = results['k_sca']
        g = results['g']
        S1 = results['S1']
        S2 = results['S2']
        lam = results['lam']

    np.savez_compressed(
        f"{location}/{name}.npz",
        a=results['a'],
        lam=lam,
        k_abs=k_abs,
        k_sca=k_sca,
        g=g,
        rho_s=results['rho_s'],
        S1=S1,
        S2=S2,
        theta=results['theta'],
    )
```

```python
def generate_and_save_opacities(f, a_grid_cm, lam, location):
    """
    Generate DSHARP opacities for a given filling factor and save them.
    """

    porosity = 1 - f

    oc, rho_s = do.get_dsharp_mix(
        porosity=porosity,
        rule="Bruggeman"
    )

    results = do.get_smooth_opacities(
        a_grid_cm,
        lam,
        rho_s,
        oc,
        extrapolate_large_grains=True
    )

    # Create the filename automatically
    filename = f"dsharp_results_ff_{f:.2f}_smoothed_extrapol"

    save_results(location, filename, results)
```

```python
def generate_and_save_opacities_not_smooth(f, a_grid_cm, lam, location):
    """
    Generate DSHARP opacities for a given filling factor and save them.
    """

    porosity = 1 - f

    oc, rho_s = do.get_dsharp_mix(
        porosity=porosity,
        rule="Bruggeman"
    )

    results = do.get_opacities(
        a_grid_cm,
        lam,
        rho_s,
        oc,
        extrapolate_large_grains=True
    )

    # Create the filename automatically
    filename = f"dsharp_results_ff_{f:.2f}_extrapol"

    save_results(location, filename, results)
```

```python
def generate_and_save_opacities_not_smooth_no_extrapolation(f, a_grid_cm, lam, location):
    """
    Generate DSHARP opacities for a given filling factor and save them.
    """

    porosity = 1 - f

    oc, rho_s = do.get_dsharp_mix(
        porosity=porosity,
        rule="Bruggeman"
    )

    results = do.get_opacities(
        a_grid_cm,
        lam,
        rho_s,
        oc,
        extrapolate_large_grains=False
    )

    # Create the filename automatically
    filename = f"dsharp_results_ff_{f:.2f}"

    save_results(location, filename, results)
```
